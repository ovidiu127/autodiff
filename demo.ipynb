```python
import numpy as np
from autodiff import Tensor
```

```python
x = Tensor([[1,2,3],[-9,-8,7]],dtype=np.float32,requires_grad=True)
y = Tensor(-2,dtype=np.float32,requires_grad=True)
```

```python
w1 = Tensor([[1,5,6]],dtype=np.float32,requires_grad=True)
w2 = Tensor([[-2,-4]],dtype=np.float32,requires_grad=True)
```

```python
z = -x + y
z = z.dot(w1.T)
z = w2.dot(z)
```

```python
z
```

```python
z.backward()
```
